```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales_data.csv")

print("Dataset loaded successfully!")
print(df)

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("TOTAL SALES:", df["Sales"].sum())
print("TOTAL PROFIT:", df["Profit"].sum())

print("\nPRODUCT-WISE SALES:")
print(df.groupby("Product")["Sales"].sum())

print("\nCATEGORY-WISE SALES:")
print(df.groupby("Category")["Sales"].sum())

print("\nTOP SELLING PRODUCT:")
print(df.groupby("Product")["Sales"].sum().idxmax())
```

```python
product_sales = df.groupby("Product")["Sales"].sum()

product_sales.plot(kind="bar", title="Product-wise Sales")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

category_sales.plot(kind="pie", autopct="%1.1f%%", title="Category-wise Sales")
plt.ylabel("")
plt.show()
```

```python
best_product = df.groupby("Product")["Sales"].sum().idxmax()
best_category = df.groupby("Category")["Sales"].sum().idxmax()

print("FINAL FINDINGS")
print("----------------------")
print("1. Total Sales:", df["Sales"].sum())
print("2. Total Profit:", df["Profit"].sum())
print("3. Best Selling Product:", best_product)
print("4. Best Performing Category:", best_category)
print("5. Total Orders:", len(df))
```
